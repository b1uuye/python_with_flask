```python
def even_odd(x):
    if x % 2 == 0:
        return 'Even'
    else:
        return 'Odd'

print(even_odd(21))
print(even_odd(8))
```

```python
def sum_list(num):
    total = 0

    for x in num:
        total += x # total = total + x
    return total
    
sum_list((1,2,3,4,5))
```
